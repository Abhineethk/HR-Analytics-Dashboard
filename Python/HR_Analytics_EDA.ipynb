```python
# Exploratory Data Analysis (EDA)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8,5))
sns.histplot(df["Age"], bins=20, kde=True)

plt.title("Distribution of Employee Age")
plt.xlabel("Age")
plt.ylabel("Count")

plt.show()
```

```python
 # Most employees are concentrated between 30–40 years of age.
# The age distribution appears approximately normal.
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df["MonthlyIncome"], bins=20, kde=True)

plt.title("Distribution of Monthly Income")
plt.xlabel("Monthly Income")
plt.ylabel("Count")

plt.show()
```

```python
# Most employees fall into lower income ranges.
# The distribution is right-skewed, with a smaller number of high-income employees.
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df["YearsAtCompany"], bins=15, kde=True)

plt.title("Distribution of Years at Company")
plt.xlabel("Years At Company")
plt.ylabel("Count")

plt.show()
```

```python
# Most employees have relatively low tenure.
# The number of employees decreases as years at company increases.
```

```python
df_corr = df.copy()

df_corr["Attrition"] = df_corr["Attrition"].map({
    "No": 0,
    "Yes": 1
})
```

```python
plt.figure(figsize=(12,8))

corr_matrix = df_corr.corr(numeric_only=True)

sns.heatmap(
    corr_matrix,
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Heatmap")

plt.show()
```

```python
# Key EDA Findings

# 1. Employees who leave earn lower salaries.
# 2. Employees who leave are younger.
# 3. Employees who leave have lower tenure.
# 4. Employees who leave report lower job satisfaction.
# 5. Overtime employees exhibit significantly higher attrition.
# 6. Most employees belong to lower salary bands.
# 7. Correlation analysis indicates relationships between salary, experience, and job level.
```
